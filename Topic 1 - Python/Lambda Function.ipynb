```python
"""
Normal Function (def)	                    Lambda Function (lambda)
Defined using def keyword.	                Defined using lambda keyword.
Has a function name.	                    Usually anonymous (no name required).
Can contain multiple statements.	        Contains only one expression.
Uses return to return a value explicitly.	Returns the expression result automatically.
Best for complex operations.	            Best for short, simple operations.
Can have a docstring.	                    Cannot have a regular function docstring.



lambda fucntion uses along with higher order function

A function that takes another function as an argument or returns a function as its result.
"""
```

```python
b=lambda x: 'Even' if x%2==0 else 'Odd'
```

```python
b(2)
```

```python
b(3)
```

```python
#Higher order function
```

```python
L = [11,14,21,23,56,78,45,29,28]
# sum of all even,odd and divisible by 3
```

```python
x=lambda x:x%2==0
y=lambda x:x%2!=0
z=lambda x:x%3==0
```

```python
def return_sum(func,L):
    result=0
    for i in L:
        if func(i):
            result+=i
    return result
```

```python
print(return_sum(x, L))
print(return_sum(y, L))
print(return_sum(z, L))
```
