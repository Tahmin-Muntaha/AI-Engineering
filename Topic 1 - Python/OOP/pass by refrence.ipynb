```python
class Customer:
    def __init__(self,name,gender):
        self.name=name
        self.gender=gender

def greet(customer): 
    ## eita function method na
    print(id(customer))
    if customer.gender =='Male':
        print('Hello',customer.name,'Sir!!')
    else:
        print('Hello',customer.name,'Madam!!')
    c2=Customer('Tahmin','Female')
    return c2
```

```python
c1=Customer('Muntaha','Female')
```

```python
print(c1.name)
```

```python
cust=greet(c1)
```

```python
print(cust.name)
```
